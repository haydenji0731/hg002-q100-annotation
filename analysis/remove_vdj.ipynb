[04/01/2026] It has come to our attention that our annotation has some VDJ segments, which should have been removed prior to Liftoff.

```python
from mjol import gan as mgan
from collections import defaultdict
import os
```

```python
to_remove_btypes = [
    'V_segment_pseudogene',
    'J_segment_pseudogene',
    'C_region',
    'C_region_pseudogene'
]
```

```python
fn = "/ccb/salz7-data/ftp.ccb/pub/data/hg002-q100/v0.5/hg002.v1.1.loff.v0.5.mat.gff"
gan = mgan.GAn(
    file_name = fn,
    file_fmt = 'gff'
)
gan.build_db()
```

```python
out_file_path = "/ccb/salz7-data/ftp.ccb/pub/data/hg002-q100/v0.6/hg002.v1.1.loff.v0.6.mat.gff"
removed = []

ctr = 0
with open(out_file_path, 'w') as fh:
    for _, f in gan.features.items():
        if f.feature_type == "gene":
            if 'gene_biotype' not in f.attributes:
                fh.write(f.to_gff_entry(include_children=True))
                ctr += 1
                continue
            btype = f.attributes['gene_biotype']
            if btype in to_remove_btypes:
                removed.append((f.attributes['ID'], f.attributes['gene_name']))
                continue
            fh.write(f.to_gff_entry(include_children=True))
            ctr += 1
ctr
```

```python
out_file_path = "/ccb/salz7-data/ftp.ccb/pub/data/hg002-q100/v0.6/removed.mat.tsv"

with open(out_file_path, 'w') as fh:
    fh.write(f'gene_id\tgene_name\n')
    for gid, gname in removed:
        fh.write(f'{gid}\t{gname}\n')
```
