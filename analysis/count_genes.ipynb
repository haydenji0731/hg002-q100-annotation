```python
from mjol import gan as mgan
from collections import defaultdict
import os
```

```python
fn = "/ccb/salz4-3/hji20/hg002-q100-annotation/results/6_add_copies/mat_loff/fmted.modified.gff"
gan = mgan.GAn(
    file_name = fn,
    file_fmt = 'gff'
)
gan.build_db()
```

```python
def is_coding_gene(f):
    for tx in f.children:
        for c in tx.children:
            if c.feature_type == "CDS":
                return True
    return False
```

```python
# count autosomal genes
counts = {
    'pc' : 0,
    'lnc' : 0,
    'r' : 0,
    'etc' : 0
}

for uid, f in gan.features.items():
    if f.feature_type == "gene":
        if 'chrX' in f.chr or 'chrY' in f.chr:
            continue
        if is_coding_gene(f):
            counts['pc'] += 1
        else:
            if 'gene_biotype' not in f.attributes:
                counts['etc'] += 1
            btype = f.attributes['gene_biotype']
            if btype == 'lncRNA':
                counts['lnc'] += 1
            elif btype == 'rRNA':
                counts['r'] += 1
            else:
                counts['etc'] += 1
counts
```

```python
sum(counts.values())
```

get collated counts using HUGO gene symbols:

```python
copy_numbers = {
    'pc' : defaultdict(int),
    'lnc' : defaultdict(int),
    'r' : defaultdict(int),
    'etc' : defaultdict(int)
}

ctr = 0
for uid, f in gan.features.items():
    if f.feature_type == "gene":
        if 'chrX' in f.chr or 'chrY' in f.chr:
            continue
        if 'gene_name' not in f.attributes:
            ctr += 1
            continue
        gname = f.attributes['gene_name']
        if 'gene_biotype' not in f.attributes:
            copy_numbers['etc'][gname] += 1
        else:
            if is_coding_gene(f):
                copy_numbers['pc'][gname] += 1
            else:
                if 'gene_biotype' not in f.attributes:
                    copy_numbers['etc'][gname] += 1
                btype = f.attributes['gene_biotype']
                if btype == 'lncRNA':
                    copy_numbers['lnc'][gname] += 1
                elif btype == 'rRNA':
                    copy_numbers['r'][gname] += 1
                else:
                    copy_numbers['etc'][gname] += 1
```

```python
copy_numbers2 = {
    'pc' : defaultdict(int),
    'lnc' : defaultdict(int),
    'r' : defaultdict(int),
    'etc' : defaultdict(int)
}

ctr = 0
for uid, f in gan2.features.items():
    if f.feature_type == "gene":
        if 'chrX' in f.chr or 'chrY' in f.chr:
            continue
        if 'gene_name' not in f.attributes:
            ctr += 1
            continue
        gname = f.attributes['gene_name']
        if 'gene_biotype' not in f.attributes:
            copy_numbers2['etc'][gname] += 1
        else:
            if is_coding_gene(f):
                copy_numbers2['pc'][gname] += 1
            else:
                if 'gene_biotype' not in f.attributes:
                    copy_numbers2['etc'][gname] += 1
                btype = f.attributes['gene_biotype']
                if btype == 'lncRNA':
                    copy_numbers2['lnc'][gname] += 1
                elif btype == 'rRNA':
                    copy_numbers2['r'][gname] += 1
                else:
                    copy_numbers2['etc'][gname] += 1
```

```python
total = 0
for k in copy_numbers:
    cn1 = copy_numbers[k]
    cn2 = copy_numbers2[k]
    union_gnames = set(cn1.keys()).union(set(cn2.keys()))
    union_cn = defaultdict(int)
    for gname in union_gnames:
        if gname in cn1 and gname in cn2:
            union_cn[gname] = max(cn1[gname], cn2[gname])
        elif gname in cn1:
            union_cn[gname] = cn1[gname]
        elif gname in cn2:
            union_cn[gname] = cn2[gname]
    print(k)
    print(sum(union_cn.values()))
    total += sum(union_cn.values())
print(total)
```
