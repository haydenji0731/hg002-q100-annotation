```python
from mjol import gan as mgan
from collections import defaultdict
```

```python
fn = "/ccb/salz4-3/hji20/hg002-q100-annotation/results/6_add_copies/mat_loff/fmted.modified.gff"
mat_gan = mgan.GAn(
    file_name = fn,
    file_fmt = 'gff'
)
mat_gan.build_db()
```

```python
fn = "/ccb/salz4-3/hji20/hg002-q100-annotation/results/6_add_copies/pat_loff/fmted.modified.gff"
pat_gan = mgan.GAn(
    file_name = fn,
    file_fmt = 'gff'
)
pat_gan.build_db()
```

```python
fn = "/ccb/salz4-3/hji20/hg002-q100-annotation/data/chm13v2.0_RefSeq_Liftoff_v5.2.gff3"
ref_gan = mgan.GAn(
    file_name = fn,
    file_fmt = 'gff'
)
ref_gan.build_db()
```

```python
status = dict()
for uid, f in ref_gan.features.items():
    if f.feature_type == "gene":
        if 'gene_biotype' in f.attributes:
            btype = f.attributes['gene_biotype']
        else:
            btype = 'None'
        if 'gene_name' in f.attributes:
            gname = f.attributes['gene_name']
            status[gname] = {
                'mat' : False, 
                'pat' : False, 
                'btype' : btype
            }
```

```python
len(status) # number of distinct gene symbols in T2T-CHM13
```

```python
for uid, f in mat_gan.features.items():
    if f.feature_type == "gene":
        if 'gene_name' in f.attributes:
            gname = f.attributes['gene_name']
            if gname in status: # if not true, those are miniprot-sourced
                status[gname]['mat'] = True
```

```python
for uid, f in pat_gan.features.items():
    if f.feature_type == "gene":
        if 'gene_name' in f.attributes:
            gname = f.attributes['gene_name']
            if gname in status: # if not true, those are miniprot-sourced
                status[gname]['pat'] = True
```

```python
out_fn = "/ccb/salz4-3/hji20/hg002-q100-annotation/analysis/missing/ann_status.tsv"

with open(out_fn, 'w') as fh:
    fh.write(f'gene_name\tbiotype\tmat\tpat\n')
    for gname in status:
        fh.write(f'{gname}\t{status[gname]['btype']}\t{status[gname]["mat"]}\t{status[gname]["pat"]}\n')
```
